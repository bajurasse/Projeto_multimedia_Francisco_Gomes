# 6. Funções embutidas nas strings

```python
texto='PNG (Portable Network Graphics) é um formato de dados utilizado para imagens,' \
    'que surgiu em 1996 como substituto para o formato GIF, devido ao facto de este último incluir algoritmos patenteados.'
lenstring=len(texto) #comprimento de uma str ou coleção de dados
print(lenstring)
numero_palavras=len(texto.split())#separa as palavras para uma lista e a len conta-as
print(numero_palavras)
findstr=texto.find('P')#indice da primeira ocorrenciaa de um elemento
print(findstr)

count=texto.count('formato')
print(count)
replace=texto.replace('formato','bajurasse')
print(replace)
```

### 6.1 Contagem de palavras de um texto(número de vezes que aparece cada palavra)

```python
texto='PNG (Portable Network Graphics) é um formato de dados utilizado para imagens,' \
    'que surgiu em 1996 como substituto para o formato GIF, devido ao facto de este último incluir algoritmos patenteados.'
palavras=texto.lower().replace ('.','').split()
frequency={}
for palavra in palavras:
    frequency[palavra]=frequency.get(palavra, 0)+1
    #já existe esta palavra? se existir diz me qtas vezes, se n da me 0
#print(frequency)
for  palavra, quantidade in frequency.items():
    print(f'{palavra}:{quantidade}')
```

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
#importamos a classe wordcloud para gerar uma nuvem de palavras e stopwords para escolher o nº de palavras q quero 
#texto='Na disciplina de multimedia vamos aprender algumas coisas interessantes'~
texto='PNG (Portable Network Graphics) é um formato de dados utilizado para imagens,' \
    'que surgiu em 1996 como substituto para o formato GIF, devido ao facto de este último incluir algoritmos patenteados.'
as_minhas_stopword= set(['Na', 'de', 'vamos', 'algumas', ])
#-------------------------configurar e gerar a nuvem de palavras a partir do texto----------------------
nuvem=WordCloud (
    width= 1000,
    height= 800,
    background_color= 'white',
    colormap= 'Greens',#Reds, viridis
    max_words=12,
    stopwords=as_minhas_stopword
).generate(texto)


#---------------------exibir a imagem atraves do modulo matplotlib------------

plt.figure(figsize=(10,5))#define o tamanho
plt.imshow(nuvem, interpolation='bilinear')# renderiza a imagem
plt.axis('off')#oculta o x e o y
plt.show()#mostra a figura
nuvem.to_file('nuvem_de_palavras.png')
```
